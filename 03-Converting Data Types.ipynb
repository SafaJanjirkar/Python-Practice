
# Converting Data Types

## Overview
Converting values between different Python data types using built-in functions.

## Topics Covered
- Converting strings to integers
- Converting lists to tuples
- Converting lists to sets
- Converting dictionary keys and values to lists

```python
num_int = 7
type(num_int)
```

```python
num_str ='7'
type(num_str)
```

```python
# This produces an error because an integer and a string cannot be added directly.
num_sum = num_int + num_str
```

```python
num_str_conv = int(num_str)
```

```python
type(num_str_conv)
```

```python
num_sum = num_int + num_str_conv
print(num_sum)
```

## Lists to Tuples

```python
num_list=[2,3,6]
type(num_list)
```

```python
tuple(num_list)
```

```python
type(tuple(num_list))
```

## Lists to Sets

Converting a list to a set removes duplicate values.

```python
num_lists2=[1,2,3,1,2,2,1,3]
```

```python
set(num_lists2)
```

```python
type(set(num_lists2))
```
