## If Elif Else Statment

```python
if 25>10:
    print("It worked..")
else:
    print("It do not worked")
```

```python
if 25<10:
    print("It worked..")
else:
    print("It do not worked")
```

```python
if 25<10:
    print("It worked..")
elif 25<20:
    print("elif 1 worked!")
elif 25<21:
    print("elif 2 worked!")
elif 25<30:
    print("elif 3 worked!")
elif 25<=30:
    print("elif 4 worked!")
else:
    print("It do not worked!")
```
