# Automatic File Sorter in File Explorer

## Overview
A Python automation project that organizes files in a folder into separate directories based on their file types.

## Features
- Creates category folders automatically
- Identifies files based on their extensions
- Moves files into the appropriate folders
- Uses Python's `os` and `shutil` modules for file management

## File Categories
- Images
- Excel files
- Documents
- PDF files

## Python Modules Used
- `os`
- `shutil`

```python
import os                    # - `os` — used to interact with the operating system and access files and folders.
import shutil                 #- `shutil` — used to perform file operations such as moving files.
```

```python
path = (r"C:/Users/YourName/Documents/AFS FE Python Project/")
```

```python
files = os.listdir(path)
```

```python
folder_names=['Image files','Excel Files','Doc files','Pdf files']
for loop in range(len(folder_names)):
 if not os.path.exists(path +folder_names[loop]):
      print(path + folder_names[loop])
      os.makedirs(path + folder_names[loop])
for file in files:
 if 'jpg' in file and not os.path.exists(path +'Image files/' + file):
     shutil.move( path +file , path +'Image files/' + file) 
     
 elif 'pdf' in file and not os.path.exists(path +'Pdf files/' + file):
     shutil.move( path +file , path +'Pdf files/' + file)
     
 elif 'doc' in file and not os.path.exists(path +'Doc files/' + file):
     shutil.move( path +file , path +'Doc files/' + file)
     
 elif 'xlsx' in file and not os.path.exists(path +'Excel files/' + file):
     shutil.move( path +file , path +'Excel files/' + file)
  
```

## Expected Result

The script automatically creates the required folders and moves files into the appropriate folder according to their file extension.

Example:

- `.jpg` → `Image files`
- `.pdf` → `Pdf files`
- `.doc` → `Doc files`
- `.xlsx` → `Excel Files`
