#### Arguments

```python
def number_squared(number):
    print(number**2)
    
```

```python
number_squared(5)
```

```python
def number_squared_custom(number,power):
    print(number**power)
```

```python
number_squared_custom(5,3)
```

```python
number_squared_custom(power= 5, number=3)
```

##### Arbitary Arguments

```python
def number_args(*number):
    print(number[0]*number[1])
```

```python
number_args(5,6,1,2,8)
```

```python
args_tuple=(5,6,1,2,8)
def number_args(*number):
    print(number[0]*number[1])
```

```python
number_args(*args_tuple)
```

#### Keyword Argument 

```python
def number_kwarg(**number):
    print('My number is: '+ number['integer'] + "My otehr number: " + number['integer2'])
```

```python
number_kwarg(integer = '1617', integer2 = '349')
```
