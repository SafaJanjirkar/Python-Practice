Understanding how functions are defined and called in Python, along with different ways of passing arguments.

## Topics Covered
- Defining and calling functions
- Function arguments
- Multiple arguments
- Keyword arguments
- Arbitrary arguments (`*args`)

A function is a reusable block of code that runs when it is called.

```python
def first_funct():
    print("We did it")
```

```python
first_funct()
```

## Function Arguments

```python
def number_squared(number):
    print(number**2)
    
```

```python
number_squared(5)
```

```python
def number_squared_custom(number,power):
    print(number**power)
```

```python
number_squared_custom(5,3)
```

```python
number_squared_custom(power= 5, number=3)
```

## Arbitrary Arguments (`*args`)

```python
def number_args(*number):
    print(number[0]*number[1])
```

```python
number_args(5,6,1,2,8)
```

```python
args_tuple=(5,6,1,2,8)
def number_args(*number):
    print(number[0]*number[1])
```

```python
number_args(*args_tuple)
```

## Keyword Arguments (`**kwargs`)

```python
def number_kwarg(**number):
    print('My number is: '+ number['integer'] + "My other number: " + number['integer2'])
```

```python
number_kwarg(integer = '1617', integer2 = '349')
```
