# BMI Calculator

## Overview
A simple Python program that calculates Body Mass Index (BMI) using weight and height entered by the user and categorizes the result based on BMI ranges.

## Features
- Accepts user input for name, weight, and height
- Calculates BMI using the provided measurements
- Categorizes the BMI result
- Uses conditional statements to determine the category

## Formula

BMI = (Weight in pounds × 703) / (Height in inches × Height in inches)


https://www.calculator.net/bmi-calculator.html

```python
Name= input("Enter your name:")
Weight= int(input("Enter your weight in pounds: "))

Height= int(input("Enter your height in inches: "))

BMI = (Weight * 703) / (Height * Height)
print(BMI)
if BMI>0 :
    if BMI<16:
        print(Name + ", you have Severe Thinness")
    elif BMI <17:
        print(Name + ", you have Moderate Thinness")
    elif BMI<18.5:
        print(Name + ", you have Mild Thinness")
    elif BMI<25:
        print(Name + ", you are Normal")
    elif BMI<30:
        print(Name + ", you are Overweight")
    elif BMI<35:
        print(Name + ", you have Obese Class I")
    elif BMI<40:
        print(Name + ", you have Obese Class II")
else: 
        print(Name + ", you have Obese Class III")

```

```python
## BMI Classification

| Category | BMI Range |
|---|---:|
| Severe Thinness | < 16 |
| Moderate Thinness | 16 - 17 |
| Mild Thinness | 17 - 18.5 |
| Normal | 18.5 - 25 |
| Overweight | 25 - 30 |
| Obese Class I | 30 - 35 |
| Obese Class II | 35 - 40 |
| Obese Class III | > 40 |
```
