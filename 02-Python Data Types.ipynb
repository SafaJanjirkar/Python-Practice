## Sets 

```python
s={1,2,3}
```

```python
type(s)
```

```python
print(s)
```

```python
daily_pint_logs={1,2,31,2,3,4,1,2,6,3,2}
print(daily_pint_logs)
```

```python
wifes_pint_logs={1,3,9,8,4,3,9,8,4,2}
print(wifes_pint_logs)
```

```python
print(daily_pint_logs | wifes_pint_logs )
```

```python
print(daily_pint_logs & wifes_pint_logs )
```

```python
print(daily_pint_logs - wifes_pint_logs )
```

```python
print( wifes_pint_logs - daily_pint_logs)
```

```python
print( wifes_pint_logs ^ daily_pint_logs)
```

## Dictionaries

```python
# Key/Value pairs

dict_name ={"name":"safa" , "weekly intake":5 , "favorite fast food":["Burger","IDK"]}
```

```python
dict_name.values()
```

```python
dict_name.keys()
```

```python
dict_name.items()
```

```python
dict_name[0]
```

```python
dict_name["name"]
```

```python
dict_name["name"]="Xyz"
print(dict_name)
```

```python
dict_name.update({"name":"safa" , "weekly intake":5 , "favorite fast food":["Burger","IDK"],"Weight":50})
print(dict_name)
```

```python
del dict_name["Weight"]
print(dict_name)
```
