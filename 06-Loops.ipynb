# Loops in Python

## Overview
Understanding how loops are used to repeatedly execute a block of code and work with collections of values.

## Topics Covered
- For loops
- Iterating through dictionaries
- Nested for loops
- While loops
- Break statements
- Continue statements

## For Loops

```python
integers =[1,2,3,4,5]
```

```python
for number in integers:
    print(number)
```

```python
for number in integers:
    print("yep")
```

```python
integers =[1,2,3,4,5]
```

```python
for i in integers:
    print(i+i)
```

```python
ice_cream_dict={"name":"Santa" , "Age": 22 , "Gender" : "Male" ,"Favorite ice cream":["Chocolate","vanilla"]}
```

```python
for cream in ice_cream_dict.values():
    print(cream)
```

```python
for key , value in ice_cream_dict.items():
    print(key," ->" , value)
```

## Nested For Loops 

```python
flavors=['Vanilla' , 'Chocolate' ,'Cookies & Cream']
toppings =["Sprinkles" , "Choco Chips" , "Oreo"]
```

```python
for one in flavors:
    for two in toppings:
        print(one,"topped with",two)
```

## While Loops

```python
number = 0

while number < 5:
    print(number)
    number = number + 1
```

## Break Statement

```python
number=0
while number <5:
 print(number)
 if number == 3: 
    break
 number= number+1
else:
    print("no longer < 5")
```

## Continue 

```python
number = 0

while number < 5:
    number = number + 1
    
    if number == 3:
        continue
    
    print(number)

else:
    print("no longer < 5")
```
