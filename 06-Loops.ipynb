# Loops 

## For Loop

```python
integers =[1,2,3,4,5]
```

```python
for number in integers:
    print(number)
```

```python
for number in integers:
    print("yep")
```

```python
integers =[1,2,3,4,5]
```

```python
for i in integers:
    print(i+i)
```

```python
ice_cream_dict={"name":"Safa" , "Age": 22 , "Gender" : "Female" ,"Favorite ice cream":["Chocolate","vanilla"]}
```

```python
for cream in ice_cream_dict.values():
    print(cream)
```

```python
for key , value in ice_cream_dict.items():
    print(key," ->" , value)
```

### Nested For Loops 

```python
flavors=['Vanilla' , 'Chocolate' ,'Cookies & Cream']
toppings =["Sprinkles" , "Choco Chips" , "Oreo"]
```

```python
for one in flavors:
    for two in toppings:
        print(one,"topped with",two)
```

## While Loops

```python
number=0
while number <5:
 print(number)
 number= number+1
```

## Break Statment

```python
number=0
while number <5:
 print(number)
 if number == 6: 
    break
 number= number+1
else:
    print("no longer < 5")
```

## Continue 

```python
number=0

while number <5:
     number= number+1
     if number == 3: 
          continue
     print(number)
else:
     print("no longer < 5")
```
